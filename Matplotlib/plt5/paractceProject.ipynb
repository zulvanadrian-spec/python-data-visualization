```python
# --- create dataframe ---
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Date":[
        "2026-01-01",
        "2026-02-01",
        "2026-03-01",
        "2026-04-01",
        "2026-05-01"
    ],
    "Products":[
        "Laptop",
        "Chair",
        "Tablet",
        "Table",
        "Keyboard"
    ],
    "Category":[
        "Electronics",
        "Furniture",
        "Electronics",
        "Furniture",
        "Electronics"
    ],
    "Price":[
        20000000,
        3000000,
        15000000,
        4000000,
        300000
    ],
    "Qty":[
        20,
        15,
        35,
        40,
        50
    ]
}

df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])

# 1.product has the highest sale

top_product = df.loc[df['Price'].idxmax()] # '.loc[df['price].idxmax()] -> ambil semua kolom index 'price', lalu '.idxmax' ambil nilai tertinggi, lalu simpan ke vsriabel
print(f"\n=== Harga Produk Tertinggi ===")
print(f"{top_product['Products']} -->  Rp.{top_product['Price']:,}")


# 2.month the highest revenue
df['Revenue'] = df['Price'] * df['Qty']
df['Month'] = df['Date'].dt.month_name()

top_month = df.loc[df['Revenue'].idxmax()] # bikin data baru "Revenue"
print(f"\n=== Revenue Bulanan Tertinggi ===")
print(f"{top_month['Month']} -->  Rp.{top_month['Revenue']:,}")

# 3.which category perform best
category_performance = df.groupby('Category')['Revenue'].sum() # ngumpilin baris yg 'Category' nya sama ('Electronics' jdi satu grup), lalu jumlahkan seluruh 'Revenue'
print(f"\n=== the Best Peerformance Category ===")
print(category_performance)

best_category = category_performance.idxmax()
print(f"The best Category : {best_category}  -->  Rp.{category_performance.max():,}")

# 4.create sales trend overtime
df.plot(
    x="Date",
    y="Revenue",
    kind="line",
    marker="o"
)
plt.title("Sales Trend Over Time (Based on Revenue)")
plt.xlabel("Date")
plt.ylabel("Revenue")

plt.show()
```

```python
# --- category comparison chart ---
category_performance.plot(
    kind="bar", 
    color=["orange", "skyblue"]
)

plt.title("Revenue Comparison by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```
