```python
# --- create data frame ---
import pandas as pd

data ={
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[120,150,180,145,210],
    "Expenses":[80,90,100,80,110]
}

df = pd.DataFrame(data)
print(df)
```

```python
# --- ploting data frame ---
import matplotlib.pyplot as plt

df.plot(
    x="Month",
    y="Sales",
    kind="line" # 'line', 'bar', 'pie'
)
plt.title("Monthly Sales Line Chart")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# --- compare Sales and Expenses ---

df.plot(
    x="Month",
    y=["Sales","Expenses"],
    kind="bar"
)
plt.title("sales vs expenses")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.show()
```

```python
# --- create a histogram ---

df["Sales"].plot(
    kind="hist",
    bins=5
)
plt.title("sales distribution")
plt.xlabel("Sales")
```

```python
# --- groupby & visualization ---
import pandas as pd2

data = {
    "Departement":[
        "Sales","sales",
        "IT","IT",
        "HR","HR"
    ],
    "Revenue":[
        100,150,
        200,180,
        80,90
    ]
}
df2 = pd2.DataFrame(data)
summary = df2.groupby("Departement")["Revenue"].sum()

print(summary)

summary.plot(kind="bar")
plt.title("Revenue by Departement")
plt.xlabel("Departement")
plt.ylabel("Revenue")

plt.show()
```

```python
# --- time series visualization ---
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Date":[
        "2026-01-01",
        "2026-02-01",
        "2026-03-01",
        "2026-04-01"
    ],
    "Sales":[100,120,160,145]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

df.plot(
    x="Date",
    y="Sales",
    kind="line"
)

plt.title("Sales Trend")
plt.xlabel("Date")
plt.ylabel("sales")

plt.show()
```
